```python
# Los proyectos con diferente estado final tienen diferencias en su valor

data['VALOR PROYECTO'] = data['VALOR PROYECTO'].astype(str).str.replace('$', '', regex=False).str.replace('€', '', regex=False).str.replace('.', '', regex=False).str.replace(',', '.', regex=False).str.replace(' ', '', regex=False).str.replace('-', '', regex=False)
data['VALOR PROYECTO'] = pd.to_numeric(data['VALOR PROYECTO'], errors='coerce')

data.groupby('ESTADO')['VALOR PROYECTO'].agg(['count', 'mean', 'median', 'min', 'max'])
```

```python
# crear el data freame

data.groupby(['ENTIDAD', 'ESTADO']).size()
```

```python
data[['ENTIDAD', 'ESTADO']].value_counts()
```

```python
#El valor promedio de los proyectos cambia dependiendo del estado final

print(data.groupby('ESTADO')['VALOR PROYECTO'].mean())
```

```python
print(data.groupby('ESTADO')['VALOR PROYECTO'].median())
```

```python
resultado_aliado = data.groupby(['ALIADO', 'ESTADO']).size()
resultado_programa = data.groupby(['PROGRAMA', 'ESTADO']).size()
resultado_convocatoria = data.groupby(['No. CONVOCATORIA', 'ESTADO']).size()

print (resultado_aliado)
print (resultado_programa)
print (resultado_convocatoria)
```
